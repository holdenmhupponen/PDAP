# Preliminaries

```r
# installing and selecting the igraph package (same as the Week Seven colab project)

install.packages('igraph')
library('igraph')
```

# Data Import

```r
# importing the data from the .csv file, as exported from KoboToolbox;
# the sep is set to ";" because the KoboToolbox csv export automatically
# separates columns by semi-colons; the check.names is set to false, because,
# I believe, it checks to see if the column names are R variables
rooms <- read.csv("koboExport.csv", header=T, sep=";", check.names=F)

# check to see if data is working
head(rooms)
```

# Extraction Operation

```r
# here I am making it so that the rooms, or the .csv export, can be extracted
# into different parts, and have those parts easily identifiable by their
# building, floor and room number; the sep is "_" so that the final ids should
# look like, Building_Floor_RoomNumber
rooms$identifier <- paste(rooms$Building, rooms$Floor, rooms$'Room Number', sep = "_")

# creating a nodes set using the identifiers created above.
# the <- should be defining "nodes" as what comes after it.
# the "[, c()]" after rooms should be indexing rooms to include multiple
# elements, which are defined afterwards
nodes <- rooms[, c(
    "node_id",
    "Building",
    "Floor",
    "Room Number",
    "Type of Room",
    "Number of Seats",
    "Types of Desks",
    "Types of Chairs",
    "Colour of Walls",
    "Colour of Ceiling"
)]

# show the nodes to check if correct
nodes
```
